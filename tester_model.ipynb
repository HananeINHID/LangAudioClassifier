```python
import numpy as np
import librosa
import tensorflow as tf
```

```python
# 1. On charge le modèle depuis ton disque dur local (Change le chemin !)
chemin_modele_local = 'modele1_routeur_meilleur.h5'
print("Chargement du modèle en local...")
modele_local = tf.keras.models.load_model(chemin_modele_local)
```

```python
# 2. La fonction d'analyse (strictement identique)
def analyser_audio_local(chemin_audio):
    y, sr = librosa.load(chemin_audio, sr=16000, mono=True)
    y_trimmed, _ = librosa.effects.trim(y, top_db=20)
    
    target_length = 16000 * 5
    if len(y_trimmed) > target_length:
        y_final = y_trimmed[:target_length]
    else:
        padding = target_length - len(y_trimmed)
        y_final = np.pad(y_trimmed, (0, padding), mode='constant')
        
    mel_spectrogram = librosa.feature.melspectrogram(y=y_final, sr=16000, n_mels=128, fmax=8000)
    mel_spectrogram_db = librosa.power_to_db(mel_spectrogram, ref=np.max)
    
    entree_modele = np.expand_dims(mel_spectrogram_db, axis=0) 
    entree_modele = np.expand_dims(entree_modele, axis=-1)     
    
    probabilites = modele_local.predict(entree_modele, verbose=0)[0]
    classe_predite = np.argmax(probabilites)
    
    labels = {0: "LOCAL", 1: "STANDARD", 2: "AUTRE"}
    
    print(f"Décision : {labels[classe_predite]}")
    return classe_predite
```

```python
# 3. Test sur un fichier audio de ton ordinateur
chemin_mon_audio = r'C:\Users\Admin\Downloads\audio55655.wav'
analyser_audio_local(chemin_mon_audio)
```
